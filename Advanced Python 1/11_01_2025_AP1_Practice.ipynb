### Write a while loop that keeps asking the user for a number until they enter a negative number. At the end, print the sum of all non-negative numbers entered.

```python
total = 0
while True:
    num = int(input('Number: '))
    if num < 0:
        break
    total += num
print(total)
```

### Using a while loop, print the first 10 Fibonacci numbers (starting from 0, 1).

```python
sec_last_num = 0
last_num = 1
count = 0
while count < 10:
    print(sec_last_num, end = " ")
    sec_last_num, last_num = last_num, sec_last_num + last_num
    count += 1
```

```python
fib = [0, 1]
while len(fib) < 10:
    fib.append(fib[-1] + fib[-2])
print(fib)
```

### You have a starting number n = 987654. Using a while loop, count how many times you can divide n by 2 (integer division) until it becomes 0.

```python
n = 987654  
steps = 0
while n > 0:
    n //= 2
    steps += 1
print(steps)
```

### You have a password "python123". Let the user try up to 3 times to enter the correct password using a while loop. Print “Access granted” or “Blocked”.

```python
pswrd = 'python123'
tries = 0
while tries < 3:
    enteredpass = input('Password: ')
    if enteredpass == pswrd:
        print('Access Granted')
        break
    else:
        print('Incorrect password!')
        tries += 1
    if tries == 3:
        print('Access Blocked')
        break

```

### Given the list nums = [3, 5, 2, 10, 6, 8], use a for loop to create a new list with only even numbers.

```python
nums = [3, 5, 2, 10, 6, 8]
evens = []
for i in nums:
    if i % 2 == 0:
        evens.append(i)
print(evens)
```

### You have a string: "advanced python practice". Count how many vowels it has using a for loop.

```python
string = "advanced python practice"
vowels = 'aeiou'
counter = 0
for ch in string:
    if ch in vowels:
        counter += 1
print(counter)
```

### Given a nested list data = [[1,2,3], [4,5], [6,7,8,9]], flatten it into a single list using a for loop.

```python
data = [[1,2,3], [4,5], [6,7,8,9]]
flat = []
for a in data:
    for b in a:
        flat.append(b)
print(flat)
```

### Define a function count_vowels(s) that returns how many vowels are in the string.

```python
def count_vowels(string):
    vowels = 'aeiouAEIOU'
    return sum(1 for i in string if i in vowels)
```

### Define a function apply_discount(price, discount=0.1) that returns the price after discount. Default discount is 10%, but user may pass another value.

```python
def apply_discount(price, discount = 0.1):  # 0.1 --> 10%
    return price * (1 - discount)
```

```python
apply_discount(2365)
```

```python
apply_discount(2365, 0.25) # discount has a default value but can be changed when calling
```

### Define a function longest_word(words) that takes a list of strings and returns the longest one. If there are multiple, return the first longest.

```python
def longest_word(list_of_words):
    longest = ''
    for word in list_of_words:
        if len(word) > len(longest):
            longest = word
    return longest
```

```python
longest_word(['hi', 'python', 'automation', 'dev'])
```
