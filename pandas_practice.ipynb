```python
# As we have null values in two different columns, while one is a categorical column, 
# the other one stores numerical data
# We must apply appropriate methods for each 
# For Age, which is a numerical column, and the age must be close to the other students, 
# we use average to fill the null value
students_df.Age.fillna(students_df.Age.mean(), inplace = True)
students_df
```

```python
# For categorical columns, if the distribution is unbalanced, 
# meaning if one value is repeated too many times eg. 100, 
# and the other categories are repeated less, then we use mode = most frequent, 
# otherwise bfill and ffill are techniques that randomly fill null values.
```

```python
students_df['Major'].fillna(students_df['Major'].mode()[0])
```
