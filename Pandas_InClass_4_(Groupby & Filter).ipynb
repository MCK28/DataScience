# <font color='green'> <b>Groupby & Aggregation Methods (aggregate() & agg())</b><font color='black'>
    
groupby(); It is used to group by a specific column in a data set.

This function is used to divide data into groups and perform various operations on these groups.

For example, a data set might include customers' names, ages, and the products they purchased. Using the groupby() function, you can group customers by the product they purchased and find their average age for each product.

[Source01](https://towardsdatascience.com/all-pandas-groupby-you-should-know-for-grouping-data-and-performing-operations-2a8ec1327b5),
[Source02](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html),
[Source03](https://naomi-fridman.medium.com/pandas-groupby-explained-with-titanic-6a1c47eb8182)
Pandas' aggregation methods perform operations on a given set of data and return a result.

These methods perform different operations such as summing the values in a data set, averaging, finding the largest or smallest value.


Some of Pandas' aggregation methods are:

**sum():** Returns the sum of all values in the specified column.

**mean():** Returns the average of all values in the specified column.

**median():** Calculates the median of grouped data.

**mode()** Returns the most recurring values in the specified column.

**max():** Returns the largest value in the specified column.

**min():** Returns the smallest value in the specified column.

**count():** Returns the total number of values in the specified column.

**std():** Returns the standard deviation of the values in the specified column.

**var():** Returns the variance of the values in the specified column.

**describe():** Calculates a statistical summary of grouped data.

**idxmin():** Returns the first index containing the minimum value along the specified axis.

**idxmax():** Returns the first index containing the maximum value along the specified axis.

**corr():** Calculates the correlation between columns, excluding NA/null values.

These methods can be used on a DataFrame or Series and operations can be performed on the desired columns or rows.

Pandas' aggregation methods are very useful for analyzing data quickly and effectively.

[Pandas Official Documentation](https://pandas.pydata.org/docs/reference/frame.html)






It is used to apply different functions (e.g. mean, sum, max) on a set of data grouped by a specific column or columns.

## <font color='blue'> <b>filter() & where()</b><font color='black'>

The **filter()** function is used to retrieve a subset of the data frame or series based on column or row labels. This is useful when you want to select specific columns or rows.

The **where()** function is used to preserve values that meet a certain condition in a data frame or series. Values that do not meet the condition are either NaN (the default) or can optionally be replaced with another value.

[Pandas Official Documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.filter.html),
[SOURCE01](https://www.sharpsightlabs.com/blog/pandas-filter/),
[SOURCE02](https://appdividend.com/2020/03/19/pandas-filter-pandas-dataframe-filter-in-python-example/),
[SOURCE03](https://www.geeksforgeeks.org/python-pandas-dataframe-where/)
    



```python
data = {
    "Company": ["Google", "Google", "Microsoft", "Microsoft", "Amazon", "Amazon", "Google", "Amazon", "Microsoft", "Google"],
    "Department": ["Engineering", "HR", "Engineering", "HR", "Engineering", "HR", "HR", "Engineering", "HR", "Engineering"],
    "Person": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah", "Ivy", "Jack"],
    "Age": [29, 35, 32, 45, 28, 38, 41, 30, 36, 27],
    "Salary": [100000, 8000, 13500, 90000, 1250, 85000, 950, 1280, 8700, 10100]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.groupby('Company').filter(lambda x : x['Age'].mean() < 35)
```

```python
df.groupby('Company').filter(lambda x : x['Age'].mean() < 35)[['Salary']]
```

```python
df.groupby('Company')[['Salary']].mean()
```

```python
df.groupby('Company').filter(lambda x : x['Salary'].mean() > 30000)
```

```python
df[df.Age < 35]
```

```python
df.where(df.Age < 35)
```

```python
df.where(df.Age < 35).dropna()
```
